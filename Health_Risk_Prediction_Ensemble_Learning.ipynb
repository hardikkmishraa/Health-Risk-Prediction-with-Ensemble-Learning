# NovaGen Research Labs

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, recall_score
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
```

# creating a baseline model

```python
df= pd.read_csv("novagen_dataset.csv")
x=df.drop("Target", axis=1)
y=df["Target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
scaler=StandardScaler()

x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
#logisitc Regression (with Regulaization)
log_reg=LogisticRegression(
    penalty='l2',
    solver='liblinear',
    max_iter=1000
)
log_reg.fit(x_train_scaled,y_train)
y_pred_lr=log_reg.predict(x_test_scaled)

#in this model evaluation, Recall is more important than accuracy
# because missing a high-risk patient is dangerous

print("Logistic Regression Accuracy:", accuracy_score(y_test,y_pred_lr))
print('Logisitc Regression recall:', recall_score(y_test,y_pred_lr))
print(classification_report(y_test,y_pred_lr))
```

# Model2 KNN

```python
knn=KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean")
knn.fit(x_train_scaled,y_train)
y_pred_knn=knn.predict(x_test_scaled)
print('knn accuracy:',accuracy_score(y_test,y_pred_knn))
print('knn recall:', recall_score(y_test,y_pred_knn))
print(classification_report(y_test,y_pred_knn))
```

# random forest ensemble

```python
rf=RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42)
rf.fit(x_train,y_train)
y_pred_rf=rf.predict(x_test)

print("random forest acc:", accuracy_score(y_test,y_pred_rf))
print("random forest recall", recall_score(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```

# Gradinet Boosting (ensemble Learning)


```python
gb=GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.1,
    max_depth=3,
    random_state=42)

gb.fit(x_train,y_train)
y_pred_gb=gb.predict(x_test)
print('gradinet acc', accuracy_score(y_test,y_pred_gb))
print('gradient recall', recall_score(y_test,y_pred_gb))
print(classification_report(y_test,y_pred_gb))
```

# voting classifier (ensemble learning)

```python
voting_clf=VotingClassifier(
    estimators=[
        ('lr', LogisticRegression(max_iter=100,solver='liblinear')),
        ('knn',KNeighborsClassifier(n_neighbors=5)),
        ('rf', RandomForestClassifier(n_estimators=200, random_state=42))
    ],
    voting='soft'
)

voting_clf.fit(x_train_scaled,y_train)
y_pred_vote=voting_clf.predict(x_test_scaled)

print("voting classififer acc:", accuracy_score(y_test,y_pred_vote))
print("voting classifier recall:", recall_score(y_test,y_pred_vote))
print(classification_report(y_test,y_pred_vote))
```
